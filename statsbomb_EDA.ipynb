# Przydatne:
1. Period (informuje o momencie meczu 1 - pierwsza połowa, 2 - druga połowa) : Int
2. position (pozycja, na której gra zawodnik wykonujący akcję) : String
3. location (pozycja strzału) : [x, y]
4. play pattern (rodzaj zagrania, z którego wyniknął strzał np. rzut wolny, atak pozycyjny) : String
5. under_pressure (informacja czy zawodnik był presowany przez obrońcę) : True/NaN
6. shot_outcome - LABEL, INFORMACJA CZY BYŁ GOL : String (np. Post, Goal)
7. shot_type (rodzaj strzału np. atak pozycyjny, rzut wolny) : String
8. shot_body_part (część ciała, którą oddano strzał) : String
9. shot_technique (technika wykonania strzału np. wolej, strzał po ziemi) : String
10. shot_first_time (czy strzał z pierwszej) : True/Nan
11. shot_deflected (czy był rykoszet) : True/Nan
12. shot_freeze_frame (koordynaty wszystkich zawodników) : lista słownikowa zawodników i lokacji
13. shot_aerial_won (czy zawodnik musiał wygrać pozycję do główki) : True/NaN

# Warte uwagi:
1. timestamp (może dodać dużo kontekstu [szarałt dla Legii]) : String
2. related_events (eventy które doprowadziły do strzału) : lista Stringów
3. shot_key_pass_id (id kluczowego podania dla danego strzału) : String/Nan

```python
el_classico_clean = el_classico_shots[['period', 'position', 'location', 'play_pattern', 'under_pressure', 'shot_outcome', 'shot_type', 'shot_body_part', 'shot_technique', 'shot_first_time', 'shot_deflected', 'shot_freeze_frame', 'shot_aerial_won', 'shot_statsbomb_xg']]
el_classico_clean
```

```python
location = el_classico_clean['location']
x_position = []
y_position = []
for location in location:
  x_position.append(location[0])
  y_position.append(location[1])

el_classico_clean.loc[:, 'location_x'] = x_position
el_classico_clean.loc[:, 'location_y'] = y_position
el_classico_clean = el_classico_clean.drop(columns=['location'])
el_classico_clean
```

```python
shot_aerial_won = (el_classico_clean['shot_aerial_won'] == True).astype(int)
shot_deflected = (el_classico_clean['shot_deflected'] == True).astype(int)
shot_first_time = (el_classico_clean['shot_first_time'] == True).astype(int)
under_pressure = (el_classico_clean['under_pressure'] == True).astype(int)
el_classico_clean.loc[:, 'shot_aerial_won'] = shot_aerial_won
el_classico_clean.loc[:, 'shot_deflected'] = shot_deflected
el_classico_clean.loc[:, 'shot_first_time'] = shot_first_time
el_classico_clean.loc[:, 'under_pressure'] = under_pressure
el_classico_clean
```

```python
is_goal = (el_classico_clean['shot_outcome'] == 'Goal').astype(int)
el_classico_clean.loc[:, 'is_goal'] = is_goal
el_classico_clean = el_classico_clean.drop(columns=['shot_outcome'])
el_classico_clean
```

```python
encoded = pd.get_dummies(el_classico_clean, columns=['position', 'play_pattern', 'shot_type', 'shot_body_part', 'shot_technique'],dtype=int, drop_first=True)
encoded
```

```python
import math

def process_players_postions(shot_freeze_frame, location_x, location_y):

  info = []
  count_enemies = 0
  count_teammates = 0
  goal_keeper_position_x = None
  goal_keeper_position_y = None
  left_post = [120, 36]
  right_post = [120, 44]

  def calculate_triangle_area(x1, y1, x2, y2, x3, y3):
    return abs((x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2)) / 2.0)

  def calculate_player_triangles_area(player_x, player_y):
    area = 0

    area += calculate_triangle_area(player_x, player_y, right_post[0], right_post[1], left_post[0], left_post[1])
    area += calculate_triangle_area(player_x, player_y, left_post[0], left_post[1], location_x, location_y)
    area += calculate_triangle_area(player_x, player_y, right_post[0], right_post[1], location_x, location_y)

    return area

  shooting_triangle_area = calculate_triangle_area(location_x, location_y, right_post[0], right_post[1], left_post[0], left_post[1])

  for player in shot_freeze_frame:
    player_area = calculate_player_triangles_area(player['location'][0], player['location'][1])
    if math.isclose(player_area, shooting_triangle_area, rel_tol=0.05) and player['teammate'] == False:
      count_enemies += 1
    if player['location'][0] >= location_x and player['teammate'] == True:
      count_teammates += 1
    if player['position']['name'] == 'Goalkeeper' and player['teammate'] == False:
      goal_keeper_position_x = player['location'][0]
      goal_keeper_position_y = player['location'][1]

  info.append(count_enemies_on_target)
  info.append(count_teammates_in_action)
  info.append(goal_keeper_position_x)
  info.append(goal_keeper_position_y)

  return info
```

```python
processed_players_locations = []
for freeze_frame, location_x, location_y in zip(encoded['shot_freeze_frame'], encoded['location_x'], encoded['location_y']):
  processed_players_locations.append(process_players_postions(freeze_frame, location_x, location_y))

columns = ['enemies_on_traget', 'teammates_up_front', 'goal_keeper_x', 'goal_keeper_y']
encoded[columns] = pd.DataFrame(processed_players_locations, index=encoded.index)

encoded.drop(columns=['shot_freeze_frame'], inplace=True)
encoded
```
